```python
"""
BEFORE MODEL (BAD PIPELINE)
--------------------------------------
Predict race-bias (1) vs non-race-bias (0)
Using raw data without cleaning or enrichment
"""

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# ===========================================
# Load the decoded but uncleaned dataset
# ===========================================

# -------------------------------------------
# 1. Target Variable (Raw Bias Category)
# -------------------------------------------
df_before["bias_raw"] = df_before["bias_1a_category"].astype(str)

df_before["target_race_bias"] = df_before["bias_raw"].apply(
    lambda x: 1 if x == "Race/Ethnicity" else 0
)

# -------------------------------------------
# 2. Basic Dirty Features
# -------------------------------------------
before_features = [
    "offense_1",
    "location_1",
    "total_victims",
    "total_offenders",
    "state_name",
    "country_region",
    "country_division"
]

df_before = df_before[before_features + ["target_race_bias"]].dropna()

# -------------------------------------------
# 3. BAD ENCODING (LabelEncoder Everything)
# -------------------------------------------
le = LabelEncoder()

for col in before_features:
    df_before[col] = df_before[col].astype(str)
    df_before[col] = le.fit_transform(df_before[col])

# -------------------------------------------
# 4. Train/Test Split
# -------------------------------------------
X = df_before[before_features]
y = df_before["target_race_bias"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# -------------------------------------------
# 5. BAD MODEL
# -------------------------------------------
model_before = LogisticRegression(max_iter=300)
model_before.fit(X_train, y_train)

y_pred_before = model_before.predict(X_test)

print("\n=== BEFORE CLEANING MODEL RESULTS ===")
print(classification_report(y_test, y_pred_before))
print(confusion_matrix(y_test, y_pred_before))
```

```python
"""
AFTER MODEL (GOOD PIPELINE)
---------------------------------------------
Predict race-bias (1) vs non-race-bias (0)
Using the fully cleaned & enriched dataset
"""

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# ===========================================
# Load cleaned + enriched dataset
# ===========================================
df = pd.read_parquet("../data/processed/hatecrimes_enriched.parquet")

# -------------------------------------------
# 1. Target Variable
# -------------------------------------------
df["target_race_bias"] = df["bias_1a_category"].apply(
    lambda x: 1 if x == "Race/Ethnicity" else 0
)

# -------------------------------------------
# 2. Feature Selection
# -------------------------------------------
after_features = [
    "offense_1",
    "location_1",
    "total_victims",
    "total_offenders",
    "victim_offender_ratio",
    "year",
    "month",
    "day_of_week",
    "is_weekend",
    "state_name",
    "state_population",
    "country_region",
    "country_division"
]

df_after = df[after_features + ["target_race_bias"]].dropna(subset=["target_race_bias"])

# Remove extreme victim/offender outliers
df_after = df_after[df_after["total_victims"] <= 20]
df_after = df_after[df_after["total_offenders"] <= 20]

# -------------------------------------------
# 3. Column Groups
# -------------------------------------------
categorical_cols = [
    "offense_1",
    "location_1",
    "day_of_week",
    "state_name",
]

numeric_cols = [
    "total_victims",
    "total_offenders",
    "victim_offender_ratio",
    "year",
    "month",
    "is_weekend",
    "state_population",
]

X = df_after[after_features].copy()
y = df_after["target_race_bias"]

# Replace pd.NA with np.nan (sklearn-safe)
X = X.replace({pd.NA: np.nan})

# -------------------------------------------
# 4. Preprocessing Pipeline
# -------------------------------------------
preprocess = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", SimpleImputer(strategy="median"), numeric_cols),
    ],
    remainder="drop"
)

# -------------------------------------------
# 5. Full Model Pipeline
# -------------------------------------------
model_after = Pipeline(
    steps=[
        ("preprocess", preprocess),
        ("clf", RandomForestClassifier(
            n_estimators=300,
            max_depth=20,
            min_samples_split=5,
            random_state=42
        ))
    ]
)

# -------------------------------------------
# 6. Train/Test Split
# -------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# -------------------------------------------
# 7. Train Model
# -------------------------------------------
model_after.fit(X_train, y_train)
y_pred_after = model_after.predict(X_test)

print("\n=== AFTER CLEANING MODEL RESULTS ===")
print(classification_report(y_test, y_pred_after))
print(confusion_matrix(y_test, y_pred_after))
```
